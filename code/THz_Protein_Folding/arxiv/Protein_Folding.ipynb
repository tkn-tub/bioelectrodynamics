```python
import urllib.request
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from prody import *

# --- Step 1: download AlphaFold structures from the GitHub server---
structures_to_fetch = {
    'AGC1': 'AF-Q12482-F1-model_v6.pdb',
    'IDHC': 'AF-P41939-F1-model_v6.pdb',
    'GABAT':'AF-P17649-F1-model_v6.pdb',
    'DHE5': 'AF-P39708-F1-model_v6.pdb',
    'FOLD': 'AF-Q12676-F1-model_v6.pdb',
    'MTAP': 'AF-Q07938-F1-model_v6.pdb',
}

base_url = "https://raw.githubusercontent.com/tkn-tub/bioelectrodynamics/main/database/"

for name, repo_path in structures_to_fetch.items():
    url = base_url + repo_path
    local_path = repo_path.split('/')[-1]
    urllib.request.urlretrieve(url, local_path)
    print(f"Downloaded {name} -> {local_path}")

# --- Step 2: unified list of everything to process ---
# Each entry: (display_label, source)
# source is either an RCSB PDB ID (str) or a local file path — parsePDB handles both
structures_to_process = [
    ('2AXQ', '2AXQ'),
    ('1YAA', '1YAA'),
    ('3FKY', '3FKY'),
    ('3IG5', '3IG5'),
] + [(name, path) for name, path in structures_to_fetch.items()]

# --- Step 3: single loop, same ANM logic as before ---
all_frequencies = []
labels = []

for label, source in structures_to_process:
    print(f"\nProcessing: {label}")
    structure = parsePDB(source)
    if structure is None:
        print(f"Warning: Could not parse {label}. Skipping.")
        continue

    calphas = structure.select('protein and name CA')
    if calphas is None or len(calphas) == 0:
        print(f"Warning: No C-alpha atoms found for {label}. Skipping.")
        continue

    anm = ANM(label)
    try:
        anm.buildHessian(calphas)
        anm.calcModes(n_modes=None)
    except Exception as e:
        print(f"Error during ANM calculation for {label}: {e}. Skipping.")
        continue

    eigvals = anm.getEigvals()
    frequencies = 0.225 * np.sqrt(eigvals)

    all_frequencies.append(frequencies)
    labels.append(f"{label} ({len(frequencies)} modes)")
    print(f"Calculated {len(frequencies)} frequencies for {label}.")

# --- Step 4: plot ---
plt.figure(figsize=(12, 6))
colors = sns.color_palette("husl", len(all_frequencies))  # auto-extends, no manual list needed

for i, freqs in enumerate(all_frequencies):
    sns.kdeplot(freqs, fill=True, color=colors[i], label=labels[i], bw_adjust=0.5)

plt.axvspan(0.1, 3.0, color='blue', alpha=0.1, label='Standard THz Range')
plt.axvline(0.1, color='blue', linestyle='--', linewidth=1)

plt.title("Vibrational Density of States (VDOS) - Experimental + AlphaFold Structures")
plt.xlabel("Frequency (THz)")
plt.ylabel("Density of Modes")
plt.xlim(0, 2.0)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()

print("VDOS plot generated for all specified structures.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming 'all_frequencies' and 'labels' are available from the previous cell (H1YaVYOPjSPp)

for i, frequencies in enumerate(all_frequencies):
    pdb_id = labels[i].split(' ')[0] # Extract PDB ID from label

    # Categorize the frequencies
    microwave_range = frequencies[frequencies < 0.1]
    thz_range = frequencies[(frequencies >= 0.1) & (frequencies <= 3.0)]
    beyond_thz_range = frequencies[frequencies > 3.0]

    print(f"\nSummary for {pdb_id}:")
    print(f"- Total vibrational modes: {len(frequencies)}")
    print(f"- Modes in GHz/Microwave range (< 0.1 THz): {len(microwave_range)}")
    print(f"- Modes in THz range (0.1 - 3.0 THz): {len(thz_range)}")
    print(f"- Modes beyond THz range (> 3.0 THz): {len(beyond_thz_range)}")


    # Plotting the distribution
    plt.figure(figsize=(10, 6))
    plt.hist(frequencies, bins=40, color='skyblue', edgecolor='black', alpha=0.7)
    plt.axvline(0.1, color='blue', linestyle='--', label='Start of THz Range (100 GHz)')
    plt.axvline(3.0, color='red', linestyle='--', label='End of THz Range (3.0 THz)')

    plt.title('Full Vibrational Frequency Distribution for {pdb_id}')
    plt.xlabel('Frequency (THz)')
    plt.ylabel('Number of Modes')
    plt.legend()
    plt.grid(axis='y', linestyle=':', alpha=0.6)
    plt.show()
```

```python
from matplotlib import animation
from IPython.display import HTML

# 1. Data Processing for 1HST
struct_1tbu = parsePDB('1TBU')
ca_1tbu = struct_1tbu.select('protein and name CA')
anm_1tbu = ANM('1TBU')
anm_1tbu.buildHessian(ca_1tbu)
anm_1tbu.calcModes(n_modes=None)
eig_1tbu = anm_1tbu.getEigvals()
freq_1tbu = 0.225 * np.sqrt(eig_1tbu)

# 3. Animation: Mode 1 (0.127 THz)
mode = anm_1tbu[0]
coords = ca_1tbu.getCoords()
vecs = mode.getArray().reshape(-1, 3)
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')
def update(i):
    ax.clear()
    curr = coords + 10.0 * np.sin(2 * np.pi * i / 20) * vecs
    ax.plot(curr[:,0], curr[:,1], curr[:,2], 'b-')
    ax.scatter(curr[:,0], curr[:,1], curr[:,2], color='red', s=5)
    ax.set_title(f'1HST Mode 1: {freq_1tbu[0]:.3f} THz')
    ax.set_axis_off()
ani = animation.FuncAnimation(fig, update, frames=20, interval=100)
ani.save('1tbu_mode1_animation.gif', writer='pillow', fps=10)
plt.close()
display(HTML(ani.to_jshtml()))
```

**Reasoning**:
Now that the ANM calculation for '1TBU' is complete, the next step is to extract the eigenvalues from the ANM model and convert them into frequencies, following the established pattern for '1HST' and '2THI'. This will allow for subsequent analysis and visualization of the vibrational modes.



```python
eigvals_1tbu = anm_1tbu.getEigvals()
f_1tbu = 0.225 * np.sqrt(eigvals_1tbu)

print(f"Calculated {len(f_1tbu)} frequencies for 1TBU, ranging from {f_1tbu.min():.3f} THz to {f_1tbu.max():.3f} THz.")
```
